```python
import feedparser
import pandas as pd
from urllib.parse import quote

START_DATE = "2022-10-01"
END_DATE   = "2025-03-30"

queries = [
    'CBDC India',
    '"Digital Rupee" RBI',
    '"e-rupee" India',
    '"Central Bank Digital Currency" India'
]

all_records = []

for q in queries:
    query = f'{q} after:{START_DATE} before:{END_DATE}'
    query_encoded = quote(query)
    rss_url = f"https://news.google.com/rss/search?q={query_encoded}&hl=en-IN&gl=IN&ceid=IN:en"

    feed = feedparser.parse(rss_url)

    for entry in feed.entries:
        all_records.append({
            "google_news_url": entry.link,
            "title": entry.title,
            "published": entry.published if "published" in entry else None,
            "source": entry.source.title if "source" in entry else None
        })

df = pd.DataFrame(all_records).drop_duplicates()

df.to_csv("google_news_cbdc_all_sites_2022_10_to_2025_03.csv", index=False)

print("====================================")
print("Total CBDC links collected:", len(df))
print("Saved file: google_news_cbdc_all_sites_2022_10_to_2025_03.csv")
print("====================================")
```

```python
import os
import re
import torch
import pandas as pd
from tqdm import tqdm
from transformers import AutoTokenizer, AutoModelForSequenceClassification

# ==========================================================
# SETTINGS
# ==========================================================
MODEL_NAME = "Moritz-Pfeifer/CentralBankRoBERTa-sentiment-classifier"
DATA_FOLDER = "news_cbdc_data"
OUTPUT_FILE = "cbdc_news_sentiment_fulltext_monthly.csv"
BATCH_SIZE = 32

# ==========================================================
# DEVICE
# ==========================================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# ==========================================================
# CBDC SENTENCE FILTER
# ==========================================================
CBDC_TERMS = [
    "cbdc", "central bank digital currency",
    "digital rupee", "e-rupee", "e₹",
    "digital currency", "digital money"
]

def is_cbdc_sentence(sentence):
    s = sentence.lower()
    return any(term in s for term in CBDC_TERMS)

_SENT_SPLIT = re.compile(r'(?<=[\.\?\!])\s+')

def split_sentences(text):
    return [s.strip() for s in _SENT_SPLIT.split(text) if len(s.strip()) > 20]

# ==========================================================
# LOAD MODEL
# ==========================================================
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME)
model.to(device).eval()

# ==========================================================
# PROCESS ONE MONTH (KEEP ALL MONTHS)
# ==========================================================
def process_month(filepath):
    fname = os.path.basename(filepath)

    # ---- Extract month (always try) ----
    match = re.search(r"(\d{4})[-_](\d{2})", fname)
    month = f"{match.group(1)}-{match.group(2)}" if match else None

    df = pd.read_excel(filepath)

    # ---- Identify text column ----
    if "TEXT" in df.columns:
        text_col = "TEXT"
    elif "text" in df.columns:
        text_col = "text"
    else:
        # ❗ KEEP MONTH, BUT NO CONTENT
        return {
            "month": month,
            "sentiment": pd.NA,
            "n_cbdc_sentences": 0
        }

    cbdc_sentences = []

    for txt in df[text_col].dropna():
        sentences = split_sentences(str(txt))
        cbdc_sentences.extend([s for s in sentences if is_cbdc_sentence(s)])

    if not cbdc_sentences:
        # ❗ KEEP MONTH, BUT NO CBDC SIGNAL
        return {
            "month": month,
            "sentiment": pd.NA,
            "n_cbdc_sentences": 0
        }

    preds_all = []

    for i in range(0, len(cbdc_sentences), BATCH_SIZE):
        batch = cbdc_sentences[i:i+BATCH_SIZE]

        inputs = tokenizer(
            batch,
            return_tensors="pt",
            padding=True,
            truncation=True,
            max_length=256
        ).to(device)

        with torch.no_grad():
            logits = model(**inputs).logits
            preds = torch.argmax(logits, dim=1)

        preds_all.extend(preds.cpu().numpy())

    n_pos = sum(p == 1 for p in preds_all)
    n_neg = sum(p == 0 for p in preds_all)
    sentiment = (n_pos - n_neg) / len(preds_all)

    return {
        "month": month,
        "sentiment": sentiment,
        "n_cbdc_sentences": len(preds_all)
    }

# ==========================================================
# RUN PIPELINE
# ==========================================================
results = []
files = sorted(f for f in os.listdir(DATA_FOLDER) if f.endswith(".xlsx"))

print("Monthly files found:", len(files))

for f in tqdm(files, desc="Processing months"):
    res = process_month(os.path.join(DATA_FOLDER, f))
    results.append(res)

monthly_df = pd.DataFrame(results)
monthly_df["month"] = pd.to_datetime(monthly_df["month"])
monthly_df = monthly_df.sort_values("month")

monthly_df.to_csv(OUTPUT_FILE, index=False)

print("\n====================================")
print("✅ CBDC NEWS SENTIMENT INDEX CREATED")
print("Saved:", OUTPUT_FILE)
print("Total months:", len(monthly_df))
print("Months with sentiment:", monthly_df['sentiment'].notna().sum())
print("====================================")
```

```python
import os
import re
import torch
import pandas as pd
from tqdm import tqdm
from transformers import AutoTokenizer, AutoModelForSequenceClassification

# ==========================================================
# SETTINGS
# ==========================================================
MODEL_NAME = "Moritz-Pfeifer/CentralBankRoBERTa-sentiment-classifier"
DATA_FOLDER = "news_cbdc_data"
OUTPUT_FILE = "cbdc_news_sentiment_paragraph_monthly.csv"
BATCH_SIZE = 32

# ==========================================================
# DEVICE
# ==========================================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# ==========================================================
# CBDC FILTER
# ==========================================================
CBDC_TERMS = [
    "cbdc", "central bank digital currency",
    "digital rupee", "e-rupee", "e₹",
    "digital currency", "digital money"
]

def is_cbdc_paragraph(paragraph):
    p = paragraph.lower()
    return any(term in p for term in CBDC_TERMS)

# ==========================================================
# PARAGRAPH SPLITTER (KEY CHANGE)
# ==========================================================
def split_paragraphs(text):
    paragraphs = re.split(r"\n\s*\n", text)
    return [p.strip() for p in paragraphs if len(p.strip()) > 100]

# ==========================================================
# LOAD MODEL
# ==========================================================
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME)
model.to(device).eval()

# ==========================================================
# PROCESS ONE MONTH
# ==========================================================
def process_month(filepath):
    fname = os.path.basename(filepath)

    # Extract YYYY-MM or YYYY_MM
    match = re.search(r"(\d{4})[-_](\d{2})", fname)
    month = f"{match.group(1)}-{match.group(2)}" if match else None

    df = pd.read_excel(filepath)

    # Accept TEXT or text
    if "TEXT" in df.columns:
        text_col = "TEXT"
    elif "text" in df.columns:
        text_col = "text"
    else:
        return {
            "month": month,
            "sentiment": pd.NA,
            "n_cbdc_paragraphs": 0,
            "n_articles": 0
        }

    article_texts = df[text_col].dropna()
    n_articles = len(article_texts)

    if n_articles == 0:
        return {
            "month": month,
            "sentiment": pd.NA,
            "n_cbdc_paragraphs": 0,
            "n_articles": 0
        }

    cbdc_paragraphs = []

    for txt in article_texts:
        paragraphs = split_paragraphs(str(txt))
        cbdc_paragraphs.extend([p for p in paragraphs if is_cbdc_paragraph(p)])

    if not cbdc_paragraphs:
        return {
            "month": month,
            "sentiment": pd.NA,
            "n_cbdc_paragraphs": 0,
            "n_articles": n_articles
        }

    preds_all = []

    for i in range(0, len(cbdc_paragraphs), BATCH_SIZE):
        batch = cbdc_paragraphs[i:i+BATCH_SIZE]

        inputs = tokenizer(
            batch,
            return_tensors="pt",
            padding=True,
            truncation=True,
            max_length=256
        ).to(device)

        with torch.no_grad():
            logits = model(**inputs).logits
            preds = torch.argmax(logits, dim=1)

        preds_all.extend(preds.cpu().numpy())

    n_pos = sum(p == 1 for p in preds_all)
    n_neg = sum(p == 0 for p in preds_all)
    sentiment = (n_pos - n_neg) / len(preds_all)

    return {
        "month": month,
        "sentiment": sentiment,
        "n_cbdc_paragraphs": len(preds_all),
        "n_articles": n_articles
    }

# ==========================================================
# RUN PIPELINE
# ==========================================================
results = []
files = sorted(f for f in os.listdir(DATA_FOLDER) if f.endswith(".xlsx"))

print("Monthly files found:", len(files))

for f in tqdm(files, desc="Processing months"):
    results.append(process_month(os.path.join(DATA_FOLDER, f)))

monthly_df = pd.DataFrame(results)
monthly_df["month"] = pd.to_datetime(monthly_df["month"])
monthly_df = monthly_df.sort_values("month")

monthly_df.to_csv(OUTPUT_FILE, index=False)

print("\n====================================")
print("✅ PARAGRAPH-LEVEL CBDC NEWS SENTIMENT CREATED")
print("Saved file:", OUTPUT_FILE)
print("Total months:", len(monthly_df))
print("Months with sentiment:", monthly_df['sentiment'].notna().sum())
print("====================================")
```
