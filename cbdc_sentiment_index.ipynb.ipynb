```python
import os
import re
import fitz
import dateparser
import torch
import pandas as pd
from tqdm import tqdm
from transformers import AutoTokenizer, AutoModelForSequenceClassification

# ---------------------------
# SETTINGS
# ---------------------------
MODEL_NAME = "Moritz-Pfeifer/CentralBankRoBERTa-sentiment-classifier"
PDF_FOLDER = r"C:\Users\Admin\Desktop\amitesh code\RBI_FINAL\newRBI_CBDC_Downloads_Final"
OUTPUT_CSV = "cbdc_sentiment_index_bis_method1.csv"
BATCH_SIZE = 50 # Number of sentences per model batch

# ---------------------------
# DEVICE (CUDA if available)
# ---------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
USE_GPU = device.type == "cuda"
print(f"Using device: {device} (USE_GPU={USE_GPU})")

# ---------------------------
# CBDC detection rules
# ---------------------------
CBDC_TERMS = [
    "cbdc", "central bank digital currency", "digital central bank money",
    "digital fiat currency", "digital fiat money", "digital base money", "dbm",
    "digital rupee", "e-money", "digital currency", "CBDC", "e₹", "e-Rupee",
    "retail cbdc", "wholesale cbdc", "retail central bank digital",
    "wholesale central bank digital", "stablecoin", "e-franc", "e-aud",
    "e-frank", "euro","e-money", "digital currency", "digital cash", "digital money",
                 "electronic currency", "electronic cash", "electronic money"
]
DIGITAL_TERMS = ["e-money", "digital currency", "digital cash", "digital money",
                 "electronic currency", "electronic cash", "electronic money"]
MODIFIERS = ["official", "national", "government", "sovereign", "legal", "public", "trial"]
CBDC_PROJECTS = ["project stella", "project dunbar", "project helvetia",
                 "project atom", "project icebreaker", "project aurora"]

def is_cbdc_sentence(sent):
    s = sent.lower()
    if any(term in s for term in CBDC_TERMS):
        return True
    if any(p in s for p in CBDC_PROJECTS):
        return True
    if any(dt in s for dt in DIGITAL_TERMS) and any(m in s for m in MODIFIERS):
        return True
    return False

# ---------------------------
# PDF helpers
# ---------------------------
def extract_text(path):
    text = ""
    with fitz.open(path) as doc:
        for page in doc:
            text += page.get_text() + "\n"
    return text

def extract_date_from_filename(path):
    """
    Extracts date from filename (YYYY-MM-DD at the start).
    Falls back to file creation date if not found.
    """
    filename = os.path.basename(path)
    match = re.match(r"(\d{4})-(\d{2})-(\d{2})", filename)
    if match:
        d = dateparser.parse(match.group(0))
        if d:
            return pd.to_datetime(d).normalize()

    # Fallback → file creation time
    try:
        return pd.to_datetime(os.path.getctime(path), unit="s").normalize()
    except Exception:
        return pd.NaT

_SENT_SPLIT = re.compile(r'(?<=[\.\?\!])\s+')
def split_sentences(text):
    return [s.strip() for s in _SENT_SPLIT.split(text) if s.strip()]

# ---------------------------
# Load model
# ---------------------------
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME).to(device).eval()

# ---------------------------
# Process one PDF with batching
# ---------------------------
def process_pdf(path):
    text = extract_text(path)
    date = extract_date_from_filename(path)
    if pd.isna(date):
        return None  # Skip if no date at all

    sentences = split_sentences(text)
    cbdc_sents = [s for s in sentences if is_cbdc_sentence(s)]

    if not cbdc_sents:
        return {"file": os.path.basename(path), "date": date, "sentiment": 0.0, "n_cbdc": 0}

    all_preds = []

    # Process in batches
    for i in range(0, len(cbdc_sents), BATCH_SIZE):
        batch_sents = cbdc_sents[i:i+BATCH_SIZE]
        inputs = tokenizer(batch_sents, return_tensors="pt", padding=True, truncation=True, max_length=512).to(device)
        with torch.no_grad():
            logits = model(**inputs).logits
            preds = torch.argmax(logits, dim=1)
            all_preds.append(preds)

    all_preds = torch.cat(all_preds)
    n_pos = (all_preds == 1).sum().item()
    n_neg = (all_preds == 0).sum().item()
    n_tot = all_preds.size(0)
    score = (n_pos - n_neg) / n_tot

    return {"file": os.path.basename(path), "date": date, "sentiment": score, "n_cbdc": n_tot}

# ---------------------------
# Build monthly index
# ---------------------------
def build_index(folder):
    start_date = pd.to_datetime("2016-01-01")
    end_date = pd.to_datetime("2025-10-30")

    files = [os.path.join(folder, f) for f in os.listdir(folder) if f.lower().endswith(".pdf")]
    rows = []

    for f in tqdm(files, desc="Processing PDFs"):
        data = process_pdf(f)
        if data is None:
            continue  # Skip PDFs with unknown dates

        # Filter by date
        if start_date <= data["date"] <= end_date:
            rows.append(data)

    if not rows:
        print("No PDFs found in the specified date range.")
        return pd.DataFrame(), pd.DataFrame()

    df = pd.DataFrame(rows)
    df["month_start"] = pd.to_datetime(df["date"]).dt.to_period("M").dt.to_timestamp()
    monthly = df.groupby("month_start")["sentiment"].mean().reset_index()
    monthly["month"] = monthly["month_start"].dt.strftime("%B")
    monthly["year"] = monthly["month_start"].dt.year
    return monthly, df

# ---------------------------
# Run script
# ---------------------------
if __name__ == "__main__":
    print("Running CBDC sentiment analysis...")

    monthly_index, doc_level = build_index(PDF_FOLDER)

    if not monthly_index.empty:
        print(monthly_index.head())
        monthly_index.to_csv(OUTPUT_CSV, index=False)
        doc_level.to_csv("1cbdc_doc_level_sentiment_bis.csv", index=False)
        print("Saved results to CSVs.")
    else:
        print("No data to save.")
```
