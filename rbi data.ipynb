```python
import undetected_chromedriver as uc
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import os
import time
import requests
import csv
from datetime import datetime
import re
import random

# --- Configuration ---
SEARCH_URL = "https://rbi.org.in/scripts/SearchResults.aspx?search=CBDC"
BASE_URL = "https://rbi.org.in"
OUTPUT_DIR = "RBI_CBDC_Downloads_Final"
REQUEST_TIMEOUT = 45
CSV_LOG_FILE = "RBI_CBDC_Download_Log.csv"

# --- Utility Functions ---
def create_directory(directory_name):
    if not os.path.exists(directory_name):
        os.makedirs(directory_name)
        print(f"✅ Directory created: '{directory_name}'")

def sanitize_filename(name, max_length=50):
    """Sanitize and truncate filename to avoid OS errors."""
    name = re.sub(r'[\\/*?:"<>|]', "", name)
    return name[:max_length].strip().replace(" ", "_")

def extract_date_from_text(text):
    """Try to extract a date from a string using multiple formats or fallback."""
    if not text:
        return "No_Date"
    for fmt in ("%b %d, %Y", "%B %d, %Y"):
        try:
            parsed_date = datetime.strptime(text, fmt)
            return parsed_date.strftime("%Y-%m-%d")
        except:
            continue
    # Fallback: try to find any 4-digit year
    match = re.search(r"\b(20\d{2})\b", text)
    if match:
        return f"{match.group(1)}"
    return "No_Date"

# --- Part 1: Scrape Links ---
def get_all_document_links(driver):
    all_page_data = []  # Store tuples of (link, heading, date)
    seen_links = set()
    page_number = 1

    print("--- Starting Part 1: Collecting links, headings, and dates ---")
    driver.get(SEARCH_URL)

    while True:
        print(f"📄 Reading Page {page_number}...")
        try:
            wait = WebDriverWait(driver, 30)
            wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "a.sub_title2_link")))
            print("   -> Content loaded successfully!")
        except Exception as e:
            print(f"   -> ⚠️ Timed out waiting for elements: {e}")
            break

        soup = BeautifulSoup(driver.page_source, 'html.parser')
        link_elements = soup.find_all('a', class_='sub_title2_link')

        if not link_elements:
            print("   -> No 'a.sub_title2_link' elements found.")
            break

        for link_tag in link_elements:
            full_link = urljoin(BASE_URL, link_tag.get('href'))
            if full_link not in seen_links:
                seen_links.add(full_link)

                heading_tag = link_tag.find('h3', class_='sub_title2')
                link_text = heading_tag.get_text(strip=True) if heading_tag else "No_Heading"

                date_tag = link_tag.find_next("span")
                date_text = date_tag.get_text(strip=True) if date_tag else None
                date_str = extract_date_from_text(date_text)

                all_page_data.append((full_link, link_text, date_str))

        print(f"   -> Found {len(all_page_data)} total items so far.")

        # Try to go to next page
        try:
            next_page_button = driver.find_element(By.LINK_TEXT, 'Next')
            driver.execute_script("arguments[0].click();", next_page_button)
            page_number += 1
            time.sleep(random.uniform(2, 4))
        except Exception:
            print("\nReached the last page. No 'Next' button found.")
            break

    return all_page_data

# --- Part 2: Download PDFs ---
def download_pdf_from_page(session, page_url, heading, date_str, csv_writer, used_names):
    """Downloads only the first PDF from a link. Ensures correct date in filename."""
    try:
        response = session.get(page_url, timeout=REQUEST_TIMEOUT)
        response.raise_for_status()
        soup = BeautifulSoup(response.content, 'html.parser')

        pdf_link_tag = soup.find('a', href=lambda href: href and href.lower().endswith('.pdf'))
        if not pdf_link_tag:
            print(f"   -> ⚠️ No PDF found on page: {page_url}")
            return False

        pdf_url = urljoin(BASE_URL, pdf_link_tag['href'])

        # Try to get a better date from the page content if date_str is "No_Date"
        if date_str == "No_Date":
            text_content = soup.get_text()
            date_str = extract_date_from_text(text_content)

        safe_heading = sanitize_filename(heading, max_length=50)

        base_name = f"{date_str}_{safe_heading}"
        counter = used_names.get(base_name, 0) + 1
        used_names[base_name] = counter
        file_name = f"{base_name}.pdf" if counter == 1 else f"{base_name}_{counter}.pdf"

        local_path = os.path.join(OUTPUT_DIR, file_name)
        if os.path.exists(local_path):
            print(f"   -> File already exists, skipping: {file_name}")
            return False

        print(f"   -> ⬇️ Downloading PDF: {file_name}")
        pdf_response = session.get(pdf_url, stream=True, timeout=REQUEST_TIMEOUT)
        pdf_response.raise_for_status()
        with open(local_path, 'wb') as f:
            for chunk in pdf_response.iter_content(chunk_size=8192):
                f.write(chunk)

        csv_writer.writerow([heading, date_str, page_url, pdf_url, file_name])
        return True

    except requests.exceptions.RequestException as e:
        print(f"   -> ❌ Failed to download from {page_url}. Error: {e}")
        return False

# --- Main Execution ---
if __name__ == "__main__":
    create_directory(OUTPUT_DIR)

    options = uc.ChromeOptions()
    options.add_argument('--headless')
    options.add_argument('--no-sandbox')
    options.add_argument('--disable-dev-shm-usage')
    options.add_argument('--disable-gpu')
    options.add_argument('--window-size=1920,1080')
    options.add_argument('--remote-debugging-port=9222')

    driver = uc.Chrome(options=options)

    # Part 1: Get all links, headings, and dates
    result_data = get_all_document_links(driver)
    driver.quit()

    print("\n" + "="*60)
    print(f"✅ Collection complete. Total unique items found: {len(result_data)}")
    print("="*60)

    if not result_data:
        print("No unique links with headings and dates were found.")
        exit()

    # Part 2: Download PDFs
    print("\n" + "="*60)
    print(f"--- Starting Part 2: Downloading PDFs ({len(result_data)} links) ---")
    print("="*60)

    session = requests.Session()
    download_count = 0
    used_names = {}

    with open(CSV_LOG_FILE, "w", newline="", encoding="utf-8") as csvfile:
        csv_writer = csv.writer(csvfile)
        csv_writer.writerow(["Heading", "Date", "Page URL", "PDF URL", "Saved Filename"])

        for i, (link, heading, date_str) in enumerate(result_data):
            print(f"Processing link {i + 1} of {len(result_data)}: {link}")
            if download_pdf_from_page(session, link, heading, date_str, csv_writer, used_names):
                download_count += 1
            time.sleep(random.uniform(1, 3))

    print("\n" + "="*60)
    print("🎉 All tasks complete!")
    print(f"Total PDFs downloaded: {download_count}")
    print(f"Files are in the '{OUTPUT_DIR}' folder.")
    print(f"Log saved to: {CSV_LOG_FILE}")
    print("="*60)
```

```python
# @title
import undetected_chromedriver as uc
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import os
import time
import requests
import csv
from datetime import datetime, timedelta
import re
import random

# --- Configuration ---
SEARCH_URL = "https://rbi.org.in/scripts/SearchResults.aspx?search=CENTRAL BANK DIGITAL CURRENCY"
BASE_URL = "https://rbi.org.in"
OUTPUT_DIR = "RBI_CBDC_Downloads_Final"
REQUEST_TIMEOUT = 60 # Increased timeout
# Define the start date for filtering
START_DATE = datetime(2016, 1, 1)


def create_directory(directory_name):
    """Creates a folder to store the downloaded files."""
    if not os.path.exists(directory_name):
        os.makedirs(directory_name)
        print(f"✅ Directory created: '{directory_name}'")

def get_all_document_links(driver):
    """
    Part 1: Uses an undetected browser to navigate pages, collect unique links and headings,
    and filter results from 2016-01-01 to today based on publication date.
    Limits scraping by the number of unique links collected.
    """
    all_page_data = []  # Store tuples of (link, heading, date)
    seen_links = set()  # Keep track of seen links to avoid duplicates
    page_number = 1
    # Calculate the date threshold (5 years ago from today)
    date_threshold = START_DATE  # Use the defined start date for filtering
    MAX_LINKS = 612  # Limit to 622 links

    print(f"--- Starting Part 1: Collecting unique links, headings, and dates (filtering from {START_DATE.strftime('%Y-%m-%d')} to today, up to {MAX_LINKS} unique links) ---")
    driver.get(SEARCH_URL)

    while len(seen_links) < MAX_LINKS:
        print(f"📄 Reading Page {page_number}...")

        # Wait for the specific link elements with class 'sub_title2_link' to be present
        try:
            wait = WebDriverWait(driver, 45)  # Increased wait time
            wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "a.sub_title2_link")))
            print("   -> Content loaded successfully!")
        except Exception as e:
            print(f"   -> ⚠️ Timed out waiting for link elements (a.sub_title2_link) to load: {e}")
            # Check if we are on the first page and found no links, then it might be the end
            if page_number == 1:
                print("   -> No links found on the first page. Assuming no results or page structure changed.")
            else:
                print("   -> Could not find expected elements on this page. Assuming end of results or an issue.")
            break

        soup = BeautifulSoup(driver.page_source, 'html.parser')

        links_on_page = []
        # Find all <a> tags with the class 'sub_title2_link'
        link_elements = soup.find_all('a', class_='sub_title2_link')

        if not link_elements:
            print("   -> No 'a.sub_title2_link' elements found on this page.")
            break  # No links found on this page, stop pagination

        for link_tag in link_elements:
            full_link = urljoin(BASE_URL, link_tag.get('href'))

            # Extract the heading text from the <h3> tag inside the link
            heading_tag = link_tag.find('h3', class_='sub_title2')
            link_text = heading_tag.get_text(strip=True) if heading_tag else "No Heading Found"

            # Find the date associated with the link
            date_span = link_tag.find_next_sibling('span', class_='search_date')
            publication_date_str = date_span.get_text(strip=True) if date_span else None
            date_str = extract_date_from_text(publication_date_str) # Ensure date_str is always set

            publication_date = None
            if date_str != "No_Date":
                 try:
                     # Attempt to parse the date string from the extracted date_str
                     publication_date = datetime.strptime(date_str, '%Y-%m-%d')
                 except ValueError:
                     print(f"   -> ⚠️ Could not parse date '{date_str}' for link: {full_link}. Using 'No_Date'.")
                     date_str = "No_Date" # Fallback if parsed date_str is still not in expected format


            # Filter by date and link limit if a valid publication date was found
            if publication_date and publication_date >= date_threshold:
                if full_link not in seen_links and len(seen_links) < MAX_LINKS:
                    seen_links.add(full_link)
                    # Always append a tuple of length 3
                    all_page_data.append((full_link, link_text, date_str))
                    links_on_page.append((full_link, link_text, date_str)) # Add to temporary list for page count
            elif publication_date and publication_date < date_threshold:
                # If a link is older than the threshold, and we found a valid date,
                # we can assume subsequent links on this page (and potentially next pages)
                # will also be older, assuming results are sorted by date.
                # This is an optimization to stop pagination early.
                print(f"   -> Found a link older than the start date ('{link_text}' dated {date_str}). Assuming end of relevant results.")
                # No need to extend all_page_data with links_on_page here as they were added incrementally
                return all_page_data  # Return the list of tuples (link, heading, date)
            else:
                # If no date was found for a link (date_str is "No_Date") or date parsing failed, include it
                if full_link not in seen_links and len(seen_links) < MAX_LINKS:
                     seen_links.add(full_link)
                     # Always append a tuple of length 3
                     all_page_data.append((full_link, link_text, date_str))
                     links_on_page.append((full_link, link_text, date_str)) # Add to temporary list for page count


        print(f"   -> Found {len(links_on_page)} new unique items (links, headings, dates) on this page (after date filtering). Current unique items: {len(all_page_data)}")


        # Stop if we have reached the maximum number of links
        if len(all_page_data) >= MAX_LINKS:
            print(f"\nReached the maximum unique item limit ({MAX_LINKS}). Stopping scraping.")
            break


        try:
            # Find the 'Next' button and click it
            next_page_button = driver.find_element(By.LINK_TEXT, 'Next')
            driver.execute_script("arguments[0].click();", next_page_button)
            page_number += 1
            time.sleep(4)  # Wait generously for the next page to load
        except Exception:
            print("\nReached the last page or no 'Next' button found.")
            break

    return all_page_data  # Return the list of tuples (link, heading, date)


def download_pdf_from_page(session, page_url, heading, date_str, csv_writer, used_names, processed_urls):
    """
    Part 2: Visits a page link, finds the PDF, and downloads it.
    Returns the page_url if no PDF is found or on download failure, otherwise returns None.
    Also checks if the URL has already been processed based on the log.
    """
    if page_url in processed_urls:
        print(f" -> Link already processed, skipping: {page_url}")
        return None # Link already processed

    try:
        response = session.get(page_url, timeout=REQUEST_TIMEOUT)
        response.raise_for_status()

        soup = BeautifulSoup(response.content, 'html.parser')
        pdf_link_tag = soup.find('a', href=lambda href: href and href.lower().endswith('.pdf'))

        if not pdf_link_tag:
            print(f" -> ⚠️ No PDF found on page: {page_url}")
            return page_url # Return URL if no PDF found

        pdf_url = urljoin(BASE_URL, pdf_link_tag['href'])
        file_name = pdf_url.split('/')[-1]
        local_path = os.path.join(OUTPUT_DIR, file_name)

        if os.path.exists(local_path):
            print(f" -> File already exists, skipping: {file_name}")
            # Even if file exists, log the URL as processed
            csv_writer.writerow([heading, date_str, page_url, pdf_url, file_name])
            processed_urls.add(page_url) # Add to processed URLs set
            return None # Return None if file exists

        print(f" -> ⬇️ Downloading PDF: {file_name}")
        pdf_response = session.get(pdf_url, stream=True, timeout=REQUEST_TIMEOUT)
        pdf_response.raise_for_status()

        with open(local_path, 'wb') as f:
            for chunk in pdf_response.iter_content(chunk_size=8192):
                f.write(chunk)

        csv_writer.writerow([heading, date_str, page_url, pdf_url, file_name])
        processed_urls.add(page_url) # Add to processed URLs set
        return None # Return None on successful download

    except requests.exceptions.RequestException as e:
        print(f" -> ❌ Failed to download from {page_url}. Error: {e}")
        return page_url # Return URL on failure


# --- Main execution block ---
if __name__ == "__main__":
    create_directory(OUTPUT_DIR)

    # Setup the undetected WebDriver with recommended options for Colab
    options = uc.ChromeOptions()
    options.add_argument('--headless') # Ensure headless mode is on
    options.add_argument('--no-sandbox') # Required for Colab
    options.add_argument('--disable-dev-shm-usage') # Overcome limited resource problems
    options.add_argument('--disable-gpu') # Disable GPU acceleration
    options.add_argument('--window-size=1920,1080') # Set a default window size
    options.add_argument('--remote-debugging-port=9222') # Add debugging port


    driver = uc.Chrome(options=options)

    # Part 1: Get all links and headings
    result_data = get_all_document_links(driver) # Capture the list of tuples
    driver.quit()

    print("\n" + "="*60)
    print(f"✅ Link and Heading collection complete. Total unique items found (filtered by date and link limit): {len(result_data)}")
    print("--- Collected Unique Data (Filtered) ---")
    if result_data:
        # Print the collected links and headings
        for link, heading, date_str in result_data:
            print(f"Heading: {heading}\nLink: {link}\nDate: {date_str}\n---")
    else:
        print("No unique links with headings were found within the specified date range and link limit.")
    print("="*60)

    # Part 2: Download PDFs (using the links from the collected data)
    download_links_data = result_data # Use the collected tuples (link, heading, date)
    links_without_pdfs = []
    processed_urls = set() # Set to keep track of URLs processed in this or previous runs

    print("\n" + "="*60)
    print(f"--- Starting Part 2: Downloading PDFs from collected links ({len(download_links_data)} links) ---")
    print("="*60)

    # Load previously processed URLs from the log file
    if os.path.exists(CSV_LOG_FILE):
        with open(CSV_LOG_FILE, "r", newline="", encoding="utf-8") as csvfile:
            reader = csv.reader(csvfile)
            header = next(reader) # Read header
            try:
                url_col_index = header.index('Page URL')
            except ValueError:
                url_col_index = -1 # Column not found

            if url_col_index != -1:
                for row in reader:
                    if len(row) > url_col_index:
                        processed_urls.add(row[url_col_index])
        print(f"Loaded {len(processed_urls)} previously processed URLs from log.")


    if download_links_data:
        session = requests.Session()
        download_count = 0
        used_names = {} # Reset used_names for this run

        # Open CSV log file in append mode
        with open(CSV_LOG_FILE, "a", newline="", encoding="utf-8") as csvfile:
             csv_writer = csv.writer(csvfile)
             # Write header only if the file is empty
             if os.path.getsize(CSV_LOG_FILE) == 0:
                 csv_writer.writerow(["Heading", "Date", "Page URL", "PDF URL", "Saved Filename"])


             for i, (link, heading, date_str) in enumerate(download_links_data):
                 print(f"Processing link {i + 1} of {len(download_links_data)}: {link}")
                 # Check if the link has already been processed based on the log
                 if link in processed_urls:
                      print(f" -> Link already processed in a previous run, skipping download attempt: {link}")
                      # Even if skipping download, increment download_count to reflect that this link is handled
                      download_count += 1
                      continue # Skip to the next link


                 no_pdf_url = download_pdf_from_page(session, link, heading, date_str, csv_writer, used_names, processed_urls)
                 if no_pdf_url is None: # PDF was downloaded, already existed, or link was processed in this run
                      download_count += 1 # Count attempts that resulted in not needing a new download
                 elif no_pdf_url is not False: # PDF was not found or download failed
                      links_without_pdfs.append((heading, no_pdf_url)) # Store heading and URL


                 time.sleep(random.uniform(1, 3))  # Random delay to avoid overloading server


        print("\n" + "="*60)
        print("🎉 All tasks complete!")
        print(f"Total PDFs downloaded or already existing (from unique links): {download_count}")
        print(f"Files are in the '{OUTPUT_DIR}' folder.")
        print(f"Log saved to: {CSV_LOG_FILE}")
        print("="*60)

        # Print links that did not have PDFs or failed to download
        if links_without_pdfs:
            print("\n--- Links without direct PDF downloads or failed downloads ---")
            for heading, link in links_without_pdfs:
                 print(f"Heading: {heading}\nLink: {link}\n---")
            print(f"\nTotal links without direct PDF downloads or failed downloads: {len(links_without_pdfs)}")
            print("="*60)
        else:
            print("\n--- All unique processed links had direct PDF downloads or downloads were attempted ---")
            print("="*60)

    else:
        print("No unique links were found to download after date filtering and link limit.")
```
